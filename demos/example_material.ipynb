```python
from xgems import ChemicalEngine, ChemicalEngineDicts, Material
```

```python
engine = ChemicalEngine("resources/mines23/mines24-dat.lst")
#enginedicts = ChemicalEngineDicts("resources/mines23/mines23-dat.lst")
#engine = ChemicalEngine("resources/cemd/CSH-Fe-dat.lst")
```

```python
enginedicts = ChemicalEngineDicts("resources/mines23/mines24-dat.lst")
# hanging
```

```python
"""Millero seawater — H2O as the solvent + dissolved ions as a dict."""
sw = Material(engine, "seawater")
sw.add("H2O", 1.00, "kg")
sw.add(
        {
            "Cl": 0.5467499,  "Na": 0.4689999,
            "Mg": 0.05282088, "Ca": 0.01037339,
            "K":  0.01020831, "S":  0.02823655,
            "C":  0.001966361,
        },
        1.0, "mol",
    )

# everything has a quantity is not given then it should be the sum of the constituents, if given then the constituents 
```

```python
"""Millero seawater — H2O as the solvent + dissolved ions as a dict."""
sw2 = Material(engine, "seawater")
sw2.add("H2O", 1.00, "kg")
sw2.add(
        {
            "Cl": 1.5467499,  "Na": 1.4689999,
            "Mg": 0.05282088, "Ca": 0.01037339,
            "K":  0.01020831, "S":  0.02823655,
            "C":  0.001966361,
        }
    )
```

```python
"""Basalt glass expressed as an oxide blend (moles per formula unit)."""
basalt = Material(engine, "basalt_glass")
for ox, mol in [
        ("SiO2",  1.000), ("Al2O3", 0.175), ("FeO",   0.169),
        ("CaO",   0.270), ("MgO",   0.260), ("Na2O",  0.040),
        ("K2O",   0.004), ("TiO2",  0.020), ("Fe2O3", 0.006),
    ]:
    basalt.add(ox, mol, "mol")
basalt
```

```python
"""Leucogranite — four-mineral assemblage by mass (per 100 g of rock)."""
granite = Material(engine, "leucogranite")
granite.add("KAlSi3O8",        17.0, "g")    # Microcline
granite.add("KAl3Si3O10(OH)2", 19.0, "g")    # Muscovite
granite.add("NaAlSi3O8",       29.0, "g")    # Albite
granite.add("SiO2",            35.0, "g")    # Quartz
```

```python
gas = Material(engine, "gas")
gas.add("CO2",        100.0, "g")
```

```python
sw_1k = sw(1, "kg")
sw_1k
```

```python
mix = sw_1k + basalt(100.0, "g") + gas
mix
```

```python
granite.set_quantity(1, 'kg')
```

```python
mix.default_amount = 1e-16
mix.b()
```

```python
engine.setColdStart()
status = engine.equilibrate(298.15, 1e5, mix.b())
print(status)
#engine
```

```python
enginedicts.set_bulk_composition(mix.b_dict())
```

```python
enginedicts.equilibrate()
```
